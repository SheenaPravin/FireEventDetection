```python
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("/Users/sheenapravin/Desktop/2025-2026/2025-2026/Consultancy/Staunch_Technologies/Marine_CCTV/Fire_Event_Detection/Fire_Event_best.pt")
```

```python
cap = cv2.VideoCapture(0)
while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = model.predict(frame_rgb)


    annotated_frame = results[0].plot()

    annotated_frame_bgr = cv2.cvtColor(annotated_frame, cv2.COLOR_RGB2BGR)

    cv2.imshow("YOLOv8 Camera Preview", annotated_frame_bgr)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
